```python
import numpy as np 
import pandas as pd
import matplotlib
import glob
```

```python

files = glob.glob("youtube-dataset/*.csv")
print(files)
```

```python
lista = []
for i in files:
    lista.append(pd.read_csv(i))
```

```python
df = pd.concat(lista ,ignore_index=True)
```

```python
X = df['CONTENT']
Y = df['CLASS']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train_,Y_test = train_test_split(X,Y , test_size=0.2, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
Vectorizer = CountVectorizer()
```

```python
X_train = Vectorizer.fit_transform(X_train)
X_test = Vectorizer.transform(X_test)
```

```python
X_train.shape
```

```python
X_train.toarray()
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
NB = MultinomialNB()
```

```python
NB.fit(X_train,Y_train_)
```

```python
NB.get_params()
```

```python
NB.feature_count_[0]
```

```python
from sklearn.metrics import ConfusionMatrixDisplay, recall_score,precision_score
```

```python
y_pred = NB.predict(X_test)
```

```python
ConfusionMatrixDisplay.from_predictions ( Y_test , y_pred )
```
