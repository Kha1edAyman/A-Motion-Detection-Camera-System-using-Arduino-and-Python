```python
import numpy as np
from numpy import random
from numpy import linalg
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
MyData = np.random.rand(100)
```

```python
MyData.shape
```

```python
MyData.reshape((20,5))
data= MyData.reshape((20,5))
```

```python
data
```

```python
print("sum  of feature 1 : " , np.sum(data ,  axis=0))
print("min  of feature 1 : " , np.min(data ,  axis=0))
print("max  of feature 1 : " , np.max(data ,  axis=0))
print("mean  of feature 1 : " , np.mean(data ,  axis=0))
print("median  of feature 1 : " , np.median(data ,  axis=0))
print("std  of feature 1 : " , np.std(data ,  axis=0))
print("var  of feature 1 : " , np.var(data ,  axis=0))
Q1 = np.quantile(data,0.25,axis=0) 
Q2 = np.quantile(data,axis=0,q=0.5) 
Q3 = np.quantile(data,axis=0,q=0.27) 
print(Q1)
print(Q2)
print(Q3)
IQR = Q3 - Q1
print(IQR)
```
